# 02 — Lidar Shape Simulation

Uji pipeline cluster → circle-fit → gate detection tanpa hardware. Bisa dimodif radius/posisi untuk menyesuaikan kondisi lapangan.

```python
from lidar import shapes
import matplotlib.pyplot as plt
import numpy as np

world = [(1400, 350, 150), (1400, -350, 150), (-2000, 1500, 600)]
angles, dists = shapes.sim_scan(world, n_beams=720)
objects, gates = shapes.detect(angles, dists)
print('objects', len(objects), 'gates', len(gates))
for o in objects: print(o)
for g in gates: print('GATE', g)
```

```python
xy = shapes.polar_to_xy(angles, dists)
fig, ax = plt.subplots()
ax.scatter(xy[:, 0], xy[:, 1], s=1, alpha=0.4)
for o in objects:
    cx,cy = o['center']; ax.add_patch(plt.Circle((cx,cy), o['diameter_mm']/2, fill=False, linewidth=1.5))
for g in gates:
    cx,cy = g['center']; ax.text(cx,cy,'GATE %.0fmm'%g['width_mm'], ha='center')
ax.axis('equal'); ax.grid(True); plt.show()
```
